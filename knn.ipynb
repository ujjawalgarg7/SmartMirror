```python
import math
import numpy as np
from collections import defaultdict
import pandas as pd

def classifyAPoint(points, p, k=3):
    # List to store (distance, group) tuples
    distance = []

    # Iterate through each group and their points
    for group in points:
        for feature in points[group]:
            # Calculate the Euclidean distance in 3D
            euclidean_distance = math.sqrt((feature[0]-p[0])**2 + (feature[1]-p[1])**2 + (feature[2]-p[2])**2)
            distance.append((euclidean_distance, group))

    # Sort the distances and get the k nearest neighbors
    distance = sorted(distance, key=lambda x: x[0])[:k]

    # Create a frequency counter for each class
    freq = defaultdict(int)
    for d in distance:
        freq[d[1]] += 1

    # Return the class with the maximum frequency
    return max(freq, key=freq.get)


# Driver function
def main():
    # Dictionary of training points having multiple labels
    points = {
        0: [(1, 12, 5), (2, 5, 8), (3, 6, 2), (3, 10, 7), (3.5, 8, 6), (2, 11, 3), (2, 9, 9), (1, 7, 4)],
        1: [(5, 3, 1), (3, 2, 8), (1.5, 9, 4), (7, 2, 2), (6, 1, 9), (3.8, 1, 7), (5.6, 4, 3), (4, 2, 5), (2, 5, 6)],
        2: [(10, 11, 1), (11, 12, 3), (9, 10, 2), (12, 13, 5)],
    }

    # Testing point p(x, y, z)
    p = (3,10,1)

    # Number of neighbors (k)
    k = 3

    # Print the result of classification for the point
    print("The value classified to unknown point is: {}".format(classifyAPoint(points, p, k)))

if __name__ == '__main__':
    main()
```
