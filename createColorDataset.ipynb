```python
import pandas as pd
```

```python
data = pd.read_csv('color_data.csv')
```

```python
data
```

```python
data = data.drop('Color_Name',axis = 1)
```

```python
data
```

```python
colors = data['Color_Family'].tolist()
colors = list(dict.fromkeys(colors))
print(colors)
```

```python
type(data)
```

```python
color_modify = []
for i in colors:
    R = 0
    G = 0
    B = 0
    C = 0
    for j in data.itertuples():
        if i == j.Color_Family:
           R = R + j.R
           G = G + j.G
           B = B + j.B
           C = C + 1
    color_modify.append([i,R/C,G/C,B/C])
```

```python
color_modify
```

```python
df = pd.DataFrame(color_modify,columns = ['Color','R','G','B'])
```

```python
df
```

```python
df.to_csv('color_dataset.csv',index=False)
```
